```python
# Importing the library
import pandas as pd
```

```python
# Loading the 1st csv file
df_sales1 = pd.read_csv("../data/Sales_MedicStore_part1.csv",sep=';')
print(df_sales1)
```

```python
# Loading the 2nd csv file
df_sales2 = pd.read_csv("../data/Sales_MedicStore_part2.csv",sep=';')
print(df_sales2)
```

```python
# Loading the excel file
df_mapping = pd.read_excel("../data/Mapping.xlsx")
print(df_mapping)
```

#Assembling

```python
# Merging the two dataframes
lst_sales = [df_sales1, df_sales2]
df_sales = pd.concat(lst_sales)
```

```python
# Displaying the first 5 rows of the merged dataframe
df_sales.head()
```

```python
# Displaying the column names of the merged dataframe
df_sales.columns
```

```python
# Displaying the first 5 rows of the mapping dataframe
df_mapping.head()
```

```python
# Displaying the column names of the mapping dataframe
df_mapping.columns
```

```python
# Merging the sales dataframe with the mapping dataframe
df_merged = pd.merge(df_sales, df_mapping)
```

```python
# Displaying the total sales of the merged dataframe
df_merged["Sales"] = df_merged["Quantity"] * df_merged["Unit Price"]
total_sales = df_merged["Sales"].sum()
print(f"Total sales: {total_sales:,.2f}")
```

```python
# Saving the merged dataframe to an excel file
df_merged.to_excel("../data/Total_Sales_MedicStore.xlsx")
```

```python
# Displaying the number of lines and columns of the merged dataframe
nb_lines_merged = df_merged.shape
print(f"Number of lines and columns: {nb_lines_merged}")
```

```python
# Displaying the number of lines and columns of the first sales dataframe
nb_lines_sales_part1 = df_sales1.shape
print(f"Number of lines and columns: {nb_lines_sales_part1}")
```

```python
# Displaying the number of lines and columns of the second sales dataframe
nb_lines_sales_part2 = df_sales2.shape
print(f"Number of lines and columns: {nb_lines_sales_part2}")
```

```python
# Displaying the number of lines and columns of the mapping dataframe
nb_lines_mapping = df_mapping.shape
print(f"Number of lines and columns: {nb_lines_mapping}")
```

```python
# Displaying the column names of the merged dataframe
df_merged.columns
```

```python
# Displaying the first 5 rows of the merged dataframe
df_merged.head(5)
```

```python
# Displaying the last 5 rows of the merged dataframe
df_merged.tail(5)
```

```python
# Displaying the summary statistics of the merged dataframe
df_merged.describe()
```

```python
# Displaying the number of unique orders in the merged dataframe
total_orders = df_merged["Order ID"].nunique()
print(f"Total number of orders: {total_orders}")
```

```python
# Displaying the number of sales in the merged dataframe
total_sales_amount = df_merged["Sales"].count()
print(f"Total sales amount: {total_sales_amount}")
```

```python
# Displaying the top 3 customers with the highest number of orders
df_merged["Customer Name"].value_counts().head(3)
```

```python
# Displaying the total sales in France
france = df_merged[df_merged["Country"] == "France"]
total_sales_france = (france["Quantity"] * france["Unit Price"]).sum()

print(f"Total sales in France: {total_sales_france:,.2f}")
```

```python
ratio_sales_france = total_sales_france / total_sales_amount * 100
print(f"France represents {ratio_sales_france:.2f}% of global sales")
```

```python
# Create a dictionary to store the analysis results


dict_analysis = {
    "How many lines are in Sales_MedicStore_part1.csv?": nb_lines_sales_part1,
    "How many lines are in Sales_MedicStore_part2.csv?": nb_lines_sales_part2,
    "How many lines are in Mapping.xlsx?": nb_lines_mapping,
    "How many orders do we have?": total_orders,
    "What is the total sales amount?": total_sales_amount,
    "What is the total Sales for France?": total_sales_france,
    "How much does the total Sales for France represent compare to the total global Sales?": ratio_sales_france
}


# Convert the dictionary to a DataFrame for better visualization
df_analysis = pd.DataFrame(list(dict_analysis.items()), columns=['Question', 'Answer'])
```

```python
#exporting the analysis results to an Excel file
df_merged.to_excel("MedicStore_Data_Analysis.xlsx", index=False)
df_analysis.head(3).to_excel("line_counts.xlsx", index=False)
df_analysis.tail(1).to_excel("stats_summary.xlsx", index=False)
df_analysis.iloc[3:6].to_excel("sales_results.xlsx", index=False)
```
