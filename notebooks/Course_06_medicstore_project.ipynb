```python
# Write your code here (import the library if you haven't already imported it)
import pandas as pd
```

```python
#loading the excel file
df_sales1 = pd.read_csv("../data/Sales_MedicStore_part1.csv",sep=';')
print(df_sales1)
```

```python
df_sales2 = pd.read_csv("../data/Sales_MedicStore_part2.csv",sep=';')
print(df_sales2)
```

```python
df_mapping = pd.read_excel("../data/Mapping.xlsx")
print(df_mapping)
```

```python
df_mapping.head(5)
```

#Assembling

```python
lst_sales = [df_sales1, df_sales2]
df_sales = pd.concat(lst_sales)
```

```python
df_sales.head()
```

```python
df_sales.columns
```

```python
df_mapping.head()
```

```python
df_mapping.columns
```

```python
df_merged = pd.merge(df_sales, df_mapping)
```

```python
df_merged["Sales"] = df_merged["Quantity"] * df_merged["Unit Price"]
total_sales = df_merged["Sales"].sum()
print(f"Total sales: {total_sales:,.2f}")
```

```python
df_merged.to_excel("../data/Total_Sales_MedicStore.xlsx")
```
